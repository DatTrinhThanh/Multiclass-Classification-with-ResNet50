```python
import gradio as gr
import tensorflow as tf
import numpy as np
import cv2
from tensorflow.keras.preprocessing.image import img_to_array
from tensorflow.keras.applications.resnet50 import preprocess_input
```

```python
model = tf.keras.models.load_model('myModel.keras') 
```

```python
def load_labels(labels_file):
    with open(labels_file, 'r') as f:
        labels = f.read().splitlines()
    return labels
```

```python
labels = load_labels('labels.txt')
```

```python
def predict_image(image):
    image = cv2.resize(image, (224, 224))
    image = img_to_array(image) 
    image = np.expand_dims(image, axis=0) 
    image = preprocess_input(image) 
    
    predictions = model.predict(image)
    predictions = predictions[0]
    
    
    result = {labels[i]: float(predictions[i]) for i in range(len(labels))}
    
    return result
```

```python
interface = gr.Interface(
    fn=predict_image,  
    inputs=gr.Image(image_mode='RGB'),
    outputs=gr.Label(num_top_classes=6),
    title="Multiclass Image Classification",
    description="Upload an image and the model will predict its class."
)

interface.launch()
```
